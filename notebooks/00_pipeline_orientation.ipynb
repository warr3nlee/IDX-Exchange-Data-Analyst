## Project mission

The internship project transforms monthly CRMLS listing and sold extracts into validated, analysis-ready housing-market data. The final program deliverables are Tableau dashboards, a one-page market intelligence report, and a presentation.

The pipeline is:

1. Obtain the available monthly extracts.
2. Generate any missing months with the approved private extraction scripts.
3. Validate coverage and schema structure.
4. Combine the monthly data.
5. Clean, enrich, and engineer market metrics.
6. Build dashboards and communicate findings.

## Confidentiality boundary

- Raw and processed MLS CSV files remain under `data/` and are excluded from Git.
- The internship handbooks and metadata documents remain private.
- Extraction scripts that reference private endpoints remain excluded from Git.
- Public code and notebooks may contain only safe aggregate counts, statistics, and visualizations.
- Addresses, personal names, contact information, coordinates, and record identifiers must never appear in saved notebook output.

## Trestle metadata fields reviewed

| Field | Working interpretation |
|---|---|
| `ListingKey` | Source-system identifier for a property-resource record |
| `PropertyType` | Broad category such as Residential, Lease, Land, or Commercial |
| `PropertySubType` | More specific category such as Single Family Residence or Condominium |
| `ListingContractDate` | Effective date of the seller-broker listing agreement |
| `PurchaseContractDate` | Date an offer was accepted and the listing left the market |
| `CloseDate` | Date the purchase agreement was fulfilled |
| `ListPrice` | Current seller/broker asking price |
| `OriginalListPrice` | Asking price on the initial listing agreement |
| `ClosePrice` | Amount paid by the purchaser |
| `LivingArea` | Total livable area within the structure |
| `DaysOnMarket` | Market time as defined by MLS business rules |
| `CountyOrParish` | County, parish, or other regional authority |
| `Latitude`, `Longitude` | Geographic reference point for the property |

These definitions guide the join keys, date-order checks, price comparisons, geographic summaries, and later feature engineering.
